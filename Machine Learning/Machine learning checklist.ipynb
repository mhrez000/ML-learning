# Machine learning

## what is machine learning? 
**Traditional programming, we write rules such as if a customer is under 25, increase their premium. In ML the cumputer looks ar past data to find those rules itself. **
*EG: Instead of guessing who might file a claim, we show a ML model thousands of past customer profiles and wether they filed a claim. The model learns the pattersn that lead to a high risk profile.*

3 types of machine learning system 

Type 1: Supervised learning = the model learns from labeled data(It knows the right answers for past examples) EG: predicting the cost of a claim based on a driver's age, car type and location.

Type 2: Unsupervised learning = The model looks for hidden patterns in the data that doesnt have lables. EG: Grouping customer with similar behaviour such sage commuters vs weekend adventurers to offer tailored plans.

Type 3: Reinforcement learning = An agent learns by trial and error to maximise a reward. EG: optimising a chatbot that helps customer file claims by rewarding it when a customer gives a thumbs up. 


The machine learning project checklist
1. Frame the problem: What is the business goal. EG: Reduce fraudulant claims by 10%

2. Get the data: Gather policy details, claim histories and external data like weather reports 

3. Explore the data: look for trends. do red cars really get into more accidents?

4. Prepare the data: This is where you clean up missing information or one hot endocode like car make into numbers the computer can understand 

5. Shortlist Models: Try different algorithms (like decision trees or linear regression)

6. Fine Tune: Tweak your best models to make thme even more accurate

7. Present: explain your finding to the insurance executives 

8. Launch: Put the model into production to start grading new applications

Data Cleaning. 🧼

Since you're aiming to be a data scientist in insurance, the method you choose depends on the specific "feature" (column) you are looking at:

Dropping Rows: If only 1% of your customers are missing their "Type of Car," you might just delete those rows. But if 40% are missing, you'd lose too much valuable data.

Imputation (Filling in):

Mean (Average): Good for data that is spread out evenly.

Median (Middle value): Better if you have "outliers." For example, if most customers have a $500 deductible but one person has a $50,000 deductible, the mean would be skewed high, but the median would stay around $500. 📈

Mode (Most frequent): Perfect for categories, like filling in a missing "Gender" or "City" with the most common one in your list.
